```python
import pandas as pd
import numpy as np
```

```python
pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)
```

```python
data = pd.read_csv("../Dataset/flats.csv")
data.head()
```

```python
# shape
data.shape
```

```python
# Identify columns with 'string' dtype
str_cols = data.select_dtypes(include='string').columns

# Convert them to object
data[str_cols] = data[str_cols].astype('object')
```

```python
# info
data.info()
```

```python
# check for duplicates
data.duplicated().sum() 
```

```python
# check for missing values
data.isnull().sum()
```

```python
# Columns to drop -> link, property_id

data.drop(columns=['link', 'property_id'], inplace=True)
data.head()
```

```python
# rename columns
data.rename(columns={'area': 'price_per_sqft'}, inplace=True)
```

```python
# Society
data['society'].value_counts()
```

```python
data['society'].nunique()
```

```python
# removing the rating from society col. and converting into lowercase 
import re
data['society'] = data['society'].apply(lambda name: re.sub(r'\d+(\.\d+)?\s?★', '', str(name)).strip()).str.lower()
data['society'].value_counts()
```

```python
# Prices -> all values are in lac or Crore, 
# except 11 properties have no values instead "Price on Request"

data['price'].value_counts()
```

```python
# remove rows having "Price per Request"
data = data[data['price'] != "Price on Request"]
data.head()
```

```python
# Converting all the values into crore

def treat_price(x):
    if type(x) == float:
        return x
    else:
        if x[1] == 'Lac':
            return round(float(x[0])/100, 2)
        else:
            return round(float(x[0]), 2)
```

```python
data['price'] = data['price'].str.split(' ').apply(treat_price)
data.head()
```

```python
# price_per_sqft -> consistent data -> all values are in /sq.fit

data['price_per_sqft'].value_counts()
```

```python
data.drop(index=2849, inplace=True)
```

```python
data.reset_index(drop=True, inplace=True)
```

```python
# data['price_per_sqft'] = (
#     data['price_per_sqft']
#     .str.extract(r'([0-9,]+)')[0]
#     .str.replace(',', '', regex=False)
#     .astype('float')
# )

data['price_per_sqft'] = (
    data['price_per_sqft']
    .str.split('/').str.get(0)
    .str.replace('₹', '', regex=False)
    .str.replace(',', '', regex=False)
    .str.strip()
    .astype('float')
)
```

```python
# bedRoom
data['bedRoom'].value_counts()
```

```python
data[data['bedRoom'].isnull()]
```

```python
# Drop all rows

data = data[~data['bedRoom'].isnull()]
data['bedRoom'].isnull().sum()
```

```python
data['bedRoom'] = data['bedRoom'].str.split(' ').str.get(0).astype('int')
data.head()
```

```python
# bathroom
data['bathroom'].value_counts()
```

```python
data['bathroom'].isnull().sum()
```

```python
data['bathroom'] = data['bathroom'].str.split(' ').str.get(0).astype('int')
data.head()
```

```python
data['balcony'].value_counts()
```

```python
# balcony
data['balcony'].isnull().sum() # 0

data['balcony'] = data['balcony'].str.split(' ').str.get(0).str.replace('No','0')
data['balcony'].value_counts()
```

```python
# additionalRoom
data['additionalRoom'].fillna('NA', inplace=True) # NA -> Not Available             
```

```python
data['additionalRoom'] = data['additionalRoom'].str.lower()
data.head()
```

```python
# floorNum
# Ground -> 0, Lower -> -1, Basement, NaN

data['floorNum'].value_counts()
```

```python
data['floorNum'].isnull().sum() # 2

data[data['floorNum'].isnull()]
```

```python

data['floorNum'] = data['floorNum'].str.split(' ').str.get(0).replace('Ground','0').str.replace('Basement','-1').str.replace('Lower','0').str.extract(r'(\d+)')
```
