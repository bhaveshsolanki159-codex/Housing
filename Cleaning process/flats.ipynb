```python
# Columns to drop -> link, property_id

data.drop(columns=['link', 'property_id'], inplace=True)
data.head()
```

```python
# rename columns
data.rename(columns={'area': 'price_per_sqft'}, inplace=True)
```

```python
# Society
data['society'].value_counts()
```

```python
data['society'].nunique()
```

```python
# removing the rating from society col. and converting into lowercase 
import re
data['society'] = data['society'].apply(lambda name: re.sub(r'\d+(\.\d+)?\s?★', '', str(name)).strip()).str.lower()
data['society'].value_counts()
```

```python
# Prices -> all values are in lac or Crore, 
# except 11 properties have no values instead "Price on Request"

data['price'].value_counts()
```

```python
# remove rows having "Price per Request"
data = data[data['price'] != "Price on Request"]
data.head()
```

```python
# Converting all the values into crore

def treat_price(x):
    if type(x) == float:
        return x
    else:
        if x[1] == 'Lac':
            return round(float(x[0])/100, 2)
        else:
            return round(float(x[0]), 2)
```

```python
data['price'] = data['price'].str.split(' ').apply(treat_price)
data.head()
```

```python
# price_per_sqft -> consistent data -> all values are in /sq.fit

data['price_per_sqft'].value_counts()
```

```python
data.drop(index=2849, inplace=True)
```

```python
data.reset_index(drop=True, inplace=True)
```
